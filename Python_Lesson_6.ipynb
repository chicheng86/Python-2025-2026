```python
def area_calculator():
    w=10
    l=50
    area=w*l
    print(area)
```

```python
area_calculator()
```

```python
def area_calculator_v1 (w,l):
    area=w*l
    print(area)
```

```python
area_calculator_v1(5,8)
```

```python
area_calculator_v1(5)
```

```python
def area_calculator_v2 (w,l):
    area=w*l
    return area
```

```python
area_calculator_v2 (3,5)
```
